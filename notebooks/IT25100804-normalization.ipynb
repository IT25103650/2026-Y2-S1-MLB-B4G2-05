```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import matplotlib.pyplot as plt
df = pd.read_csv("output_after_encoding.csv")
numeric_cols = ["ticket_price", "spend_amount"]
df[numeric_cols].describe()
```

```python
# Step 1: Demonstrate why scaling matters

ranges = df[numeric_cols].agg(["min", "max"]).T
ranges["range"] = ranges["max"] - ranges["min"]
ranges
```

```python
# Step 2: Apply StandardScaler (z-score) -

scaler = StandardScaler()
df_scaled = df.copy()
df_scaled[numeric_cols] = scaler.fit_transform(df[numeric_cols])

df_scaled[numeric_cols].describe().round(2)
```

```python
# Step 3 (comparison): MinMaxScaler alternative 

minmax_scaler = MinMaxScaler()
df_minmax = df.copy()
df_minmax[numeric_cols] = minmax_scaler.fit_transform(df[numeric_cols])
df_minmax[numeric_cols].describe().round(2)
```

## Generating output

```python
df_scaled.to_csv("output_after_scaling.csv", index=False)
print("Saved output_after_scaling.csv - shape:", df_scaled.shape)
```
